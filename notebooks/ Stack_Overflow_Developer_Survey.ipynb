To understand the central tendency of numerical features, I will first identify all numerical columns in the results_public DataFrame. Then, I will calculate and display the mean, median, and mode for each of these columns as instructed.

```python
numerical_cols = public_copy.select_dtypes(include=np.number)

print("Measures of Central Tendency for Numerical Columns:\n")

for col in numerical_cols.columns:
    mean_val = numerical_cols[col].mean()
    median_val = numerical_cols[col].median()
    mode_val = numerical_cols[col].mode()

    print(f"Column: {col}")
    print(f"  Mean:   {mean_val:.2f}")
    print(f"  Median: {median_val:.2f}")

    # Mode can return multiple values, so print all if there are multiple
    if len(mode_val) > 1:
        print(f"  Mode:   {mode_val.tolist()}")
    elif not mode_val.empty:
        print(f"  Mode:   {mode_val[0]:.2f}")
    else:
        print("  Mode:   No unique mode or all values are unique.")
    print("\n")
```

I will calculate the percentage of missing values for each column in the results_public DataFrame and display them to understand the extent of missing data.

```python
missing_percentages = public_copy.isnull().sum() * 100 / len(public_copy)
print("Percentage of missing values per column:")
print(missing_percentages[missing_percentages > 0].sort_values(ascending=False))
```

I replace missing values ​​in numeric columns with the median.

```python
numerical_cols = public_copy.select_dtypes(include=np.number).columns
for col in numerical_cols:
    if public_copy[col].isnull().any():
        median_val = public_copy[col].median()
        public_copy[col].fillna(median_val, inplace=True)
        print(f"Filled missing numerical values in column '{col}' with median: {median_val:.2f}")
```

First, I will filter the 'results_public' DataFrame to include only respondents who are '25-34 years old' and store it in df_age_filtered. Then, I will further filter this DataFrame to include only those with a 'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)' for their 'EdLevel'.

```python
df_age_filtered = public_copy[public_copy['Age'] == '25-34 years old']
df_master_age_filtered = df_age_filtered[df_age_filtered['EdLevel'] == 'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)']

print("First 5 rows of df_master_age_filtered:")
print(df_master_age_filtered.head())
print("\nShape of df_master_age_filtered:")
print(df_master_age_filtered.shape)
```

I will calculate the mean of 'JobSat' (Job Satisfaction) by 'EdLevel'.

```python
mean_jobsat_by_edlevel = public_copy.groupby('EdLevel')['JobSat'].mean().sort_values(ascending=False)
print("Mean Job Satisfaction by Educational Level (since 'ConvertedCompYearly' was dropped):")
print(mean_jobsat_by_edlevel)
```

I will select an existing relevant categorical column, 'AIHuman', which contains semi-colon separated values, and extract all unique individual values from it to understand the different human-AI interaction scenarios mentioned by respondents.

```python
all_ai_human_interactions = set()
for entry in public_copy['AIHuman'].dropna():
    interactions = entry.split(';')
    for interaction in interactions:
        all_ai_human_interactions.add(interaction.strip())

print("Unique individual human-AI interaction scenarios ('AIHuman' column):")
print(sorted(list(all_ai_human_interactions)))
```

To apply the complex filter, I will combine the three specified conditions ('Age' == '25-34 years old', 'Employment' == 'Employed', and 'EdLevel' == 'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)') using logical AND operators.

```python
df_complex_filtered = public_copy[
    (public_copy['Age'] == '25-34 years old') &
    (public_copy['Employment'] == 'Employed') &
    (public_copy['EdLevel'] == 'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)')
]
print("First 5 rows of df_complex_filtered:")
print(df_complex_filtered.head())
print("\nShape of df_complex_filtered:")
print(df_complex_filtered.shape)
```

To visualize the mean job satisfaction by educational level, I will create a bar plot using the mean_jobsat_by_edlevel Series. This will clearly show how job satisfaction varies across different educational backgrounds.

**Task 9.** Analysis of Python popularity by age group.

```python
python_developers_df = public_copy[public_copy['LanguageHaveWorkedWith'].fillna('').str.contains('Python', na=False)]

total_developers_by_age = public_copy['Age'].value_counts()
print("Total developers by age group:")
print(total_developers_by_age)

python_developers_by_age = python_developers_df['Age'].value_counts()
print("\nPython developers by age group:")
print(python_developers_by_age)

percentage_python_by_age = (python_developers_by_age / total_developers_by_age) * 100
percentage_python_by_age = percentage_python_by_age.fillna(0).sort_index()

print("\nPercentage of Python developers by age group:")
print(percentage_python_by_age)
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x=percentage_python_by_age.index, y=percentage_python_by_age.values, palette='viridis', hue=percentage_python_by_age.index, legend=False)
plt.xlabel('Age Group')
plt.ylabel('Percentage of Python Developers')
plt.title('Percentage of Python Developers by Age Group')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--')
plt.tight_layout()
plt.show()
```

### **BUSINESS INSIGHTS AND RECOMMENDATIONS**


**1. Demographics and Target Audience**

**Insights:**
*   The most active and strategically important group of developers is aged 25–34.
*   This age group has a high share of:
1.   individuals with a Master’s degree,
2.   full-time employed professionals,
3.   developers actively using Python.
*   This segment represents a core part of the IT talent market.

**Business Recommendations:**
*   Focus products, services, and marketing efforts on the 25–34 age group.
*   Emphasize more “mature” value propositions such as:
1.   income growth,
2.   career development,
3.   work-life balance.





**2. Education and Job Satisfaction**

**Insights:**
*   Job satisfaction is higher among professionals with higher education, especially those holding a Master’s degree.
*   Education correlates not only with salary levels but also with employee loyalty.

**Business Recommendations:**
*   Invest in:
1.   corporate training programs,
2.   compensation for Master’s degrees and professional courses,
3.   upskilling and reskilling initiatives.
*   Use education as a retention tool, not only as a hiring mechanism.












**3. Remote Work**

**Insights:**
*   A significant proportion of professionals work fully remotely.
*   Among remote workers, there is a high-income segment (above the 75th percentile of earnings).

**Business Recommendations:**
*   Remote work is not a compromise, but a competitive advantage.
*   For highly paid specialists, remote work is a standard expectation, not a perk.
*   Companies without a remote work policy risk losing access to top-tier talent.

**4. Python as a Strategic Language**

**Insights:**
*   Python is one of the most popular programming languages.
*   It is used by:
1.   a large share of developers,
2.   especially actively by professionals aged 25–44.
*   Python developers are present across all regions and industries.

**Business Recommendations:**
*   Make Python a core or priority technology stack for:
1.   data analytics,
2.   automation,
3.   AI/ML,
4.   backend development.
*   Training in Python is a fast and efficient way to scale teams.
*   Python helps reduce time-to-market.
















**5. Learning and Skill Development Channels**

**Insights:**
*   Online courses are one of the most popular learning methods.
*   Formal education is increasingly complemented, not replaced, by online learning.

**Business Recommendations:**
*   Leverage:
1.   subscriptions to online learning platforms,
2.   internal training programs,
3.   hybrid learning formats.
*   These approaches are more cost-effective than hiring and help close skill gaps faster.










**6. Python Developers’ Salaries by Country**

**Insights:**
*   There is a significant salary variation across countries.
*   Mean and median compensation differ substantially, indicating the presence of outliers.
*   Geography has a strong impact on the cost of talent.

**Business Recommendations:**
*   Use global hiring strategies:
1.   recruit in regions with a strong price-to-quality ratio,
2.   while maintaining competitive compensation levels.
*   Optimize payroll costs without compromising talent quality.









**7. Highest-Paid Specialists**

**Insights:**
*   The top 5 highest-paid respondents are:
1.   predominantly individuals with higher education,
2.   often holding a Master’s degree.
*   High compensation results from a combination of:
1.   education,
2.   experience,
3.   in-demand technology stacks.

**Business Recommendations:**
*   For critical roles:
1.   do not compromise on candidate quality,
2.   invest in A-players.
*   A cheaper specialist is not necessarily more cost-effective in the long term.








**8. Age and Python Adoption**

**Insights:**
*   Python is most popular among:
1.   younger professionals,
2.   mid-level specialists.
*   Python is a long-term technology, not a temporary trend.

**Business Recommendations:**
*   Prioritize Python in:
1.   long-term product planning,
2.   technical strategy development.
*   Reduce the risk of technology stack obsolescence.










**9. Industries with Highly Paid Remote Specialists**

**Insights:**
*   Top industries include:
1.   IT products,
2.   finance,
3.   consulting,
4.   SaaS.
*   These industries are the most willing to pay premium compensation for remote talent.

**Business Recommendations:**
*   If a company operates outside these industries:
1.   strengthen the EVP (Employee Value Proposition),
2.   or compensate with bonuses, flexibility, and career growth opportunities.
